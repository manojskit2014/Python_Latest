```python
# print odd numbers in the list
# (20,10,16,19,25,1,276,188)
list = eval(input('Enter the list : '))
print(list)
print(type(list))
for each_element in list:
    if each_element %2 != 0:
        print(each_element, "is odd")
```

```python
# print Even numbers in the list
(20,10,16,19,25,1,276,188)
list = eval(input('Enter the list : '))
print(list)
print(type(list))
for each_element in list:
    if each_element %2 == 0:
        print(each_element, "is even")
```
